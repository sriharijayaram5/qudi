```python
import numpy as np

# Example dimensions
num_runs = 2
freq_points = 4
sweeps_per_run = 2
n_bins = 2

# Example data
arr = np.random.rand(num_runs, freq_points * sweeps_per_run, n_bins)
print('Init array:')
print(arr)

# Reshape to expose freq_points dimension
arr_reshaped = arr.reshape(num_runs,sweeps_per_run , freq_points, n_bins)
print('Reshaped array:')
print(arr_reshaped)

# Sum over all but freq_points
result = arr_reshaped.sum(axis=(0, 2, 3))
print('Result array:')
print(result)

# print(result.shape)  # (10,)
```

```python
import numpy as np

# Example dimensions
num_runs = 2
freq_points = 4
sweeps_per_run = 2
n_bins = 3

# Example data
arr = np.random.rand(num_runs, freq_points * sweeps_per_run, n_bins)
print('Init array:')
print(arr)

# Step 1: reshape to separate freq_points and sweeps_per_run
arr_reshaped = arr.reshape(num_runs*sweeps_per_run,freq_points, n_bins)
print('Reshaped array:')
print(arr_reshaped)

# Step 2: sum over runs and sweeps
result = arr_reshaped[0:2,:,:].sum(axis=(0))
print('Result array:')
print(result)

# print(result.shape)  # (freq_points, n_bins)
```
