```python
import numpy as np

# Parameters
shape = (100, 100)       # Desired shape
prob_true = 0.3        # Proportion of True values (between 0 and 1)

# Compute number of elements and number of True values
total_elements = np.prod(shape)
num_true = int(round(prob_true * total_elements))

# Create array with exact number of Trues and Falses
bools = np.array([True] * num_true + [False] * (total_elements - num_true))

# Shuffle and reshape
np.random.shuffle(bools)
random_bools = bools.reshape(shape)
print(random_bools)
```
