```python
import random
import numpy as np
import matplotlib.pyplot as plt
```

```python
for idy, coord_y in enumerate(afm_scanner_logic._pulsed_scan_array['pulsed_fw']['coord1_arr']):
    for idx, coord_x in enumerate(afm_scanner_logic._pulsed_scan_array['pulsed_fw']['coord0_arr']):
        for id_var, var in enumerate(afm_scanner_logic._pulsed_scan_array['pulsed_fw']['coord2_arr']):
            afm_scanner_logic._pulsed_scan_array['pulsed_fw']['data'][idy][idx][id_var] = random.uniform(0.1,5)
            afm_scanner_logic._pulsed_scan_array['pulsed_fw']['data_alternating'][idy][idx][id_var] = random.uniform(0.1,5)
            afm_scanner_logic._pulsed_scan_array['pulsed_fw']['data_delta'][idy][idx][id_var] = random.uniform(0.1,5)
```
