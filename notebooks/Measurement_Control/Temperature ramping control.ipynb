```python
import numpy as np
import time
from tqdm import tqdm
```

### Define setpoint ramping parameters

```python
sample_temp_goal_setpoint = 82 #in Kelvin
vti_temp_goal_setpoint = 3 #in Kelvin
temperature_accuracy = 0.01 #in Kelvin
ramping_minimum_timestep = 3 #in seconds
ramping_points = 100
```

### Ramping sample temperature setpoint

```python
attodry2200.jupyter_ramping_control_stop = False
force_break = False
current_sample_temp = attodry2200.get_sample_temp()
if current_sample_temp>sample_temp_goal_setpoint:
    setpoint_arr = np.linspace(current_sample_temp, sample_temp_goal_setpoint, ramping_points) #temperature increase
else:
    setpoint_arr = np.linspace(sample_temp_goal_setpoint, current_sample_temp, ramping_points) #temperature decrease

attodry2200.start_sample_temp_control()
for setpoint in tqdm(setpoint_arr):
    attodry2200.set_sample_temp_setpoint(setpoint)
    time.sleep(ramping_minimum_timestep)
    while True:
        current_temp = attodry2200.get_sample_temp()
        if current_temp == -1:
            force_break = True
            break
        if (abs(setpoint - current_temp) < temperature_accuracy) or attodry2200.jupyter_ramping_control_stop:
            break
        time.sleep(ramping_minimum_timestep*0.1)
    if force_break:
        print('Temperature sweep stopped due to sensor issue!')
        break
    if attodry2200.jupyter_ramping_control_stop:
        break
        
```

### Ramping VTI temperature setpoint

```python
attodry2200.jupyter_ramping_control_stop = False
force_break = False
current_vti_temp = attodry2200.get_vti_temp()
if current_vti_temp>vti_temp_goal_setpoint:
    setpoint_arr = np.linspace(current_vti_temp, vti_temp_goal_setpoint, ramping_points) #temperature increase
else:
    setpoint_arr = np.linspace(vti_temp_goal_setpoint, current_vti_temp, ramping_points) #temperature decrease

attodry2200.start_vti_temp_control()
for setpoint in tqdm(setpoint_arr):
    attodry2200.set_vti_temp_setpoint(setpoint)
    time.sleep(ramping_minimum_timestep)
    while True:
        current_temp = attodry2200.get_vti_temp()
        if current_temp == -1:
            force_break = True
            break
        if (abs(setpoint - current_temp) < temperature_accuracy) or attodry2200.jupyter_ramping_control_stop:
            break
        time.sleep(ramping_minimum_timestep*0.1)
    if force_break:
        print('Temperature sweep stopped due to sensor issue!')
        break
    if attodry2200.jupyter_ramping_control_stop:
        break
    
```
